```python
import pandas as pds
```

```python
pwd
```

```python
df2 = pds.read_csv("student_lifestyle_dataset.csv")
```

```python
df2.head()
```

```python
total = df2.groupby('gender')['sleep_hours'].sum()
```

```python
#Females have more sleep hours
total
```

```python
plt.scatter('screen_time', 'sleep_hours' , data=df2 , c='blue', s=50, alpha=1)
plt.xlabel('screen_time')
plt.ylabel('sleep_hours')
plt.title('Scatter Plot')
plt.show ()
#There is no correlation
```

```python
plt.hist(df2['daily_transport_cost'], bins=20, color='blue', edgecolor='black', alpha=0.7)

plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.title('Histogram of Daily Transport Cost')
plt.show()
```

```python
avg_internet = df2.groupby('gender')['internet_usage'].mean()
```
